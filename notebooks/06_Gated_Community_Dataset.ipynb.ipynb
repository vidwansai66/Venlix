```python
import pandas as pd
import numpy as np

# -----------------------------
# Load Dataset
# -----------------------------
df = pd.read_csv("../data/gated_delivery_dataset.csv")

np.random.seed(42)

# -----------------------------
# Customer Answered Call
# -----------------------------
df["customer_answered_call"] = np.random.choice(
    ["Yes", "No"],
    size=len(df),
    p=[0.82, 0.18]
)

# -----------------------------
# Customer Response Time (seconds)
# -----------------------------
response_time = []

for call in df["customer_answered_call"]:
    if call == "Yes":
        response_time.append(np.random.randint(5, 120))
    else:
        response_time.append(np.random.randint(180, 600))

df["customer_response_time"] = response_time

# -----------------------------
# Customer Availability
# -----------------------------
df["customer_availability"] = np.random.choice(
    ["Home", "Office", "Travelling", "Unknown"],
    size=len(df),
    p=[0.45, 0.25, 0.20, 0.10]
)

# -----------------------------
# Visitor Pass Status
# -----------------------------
df["visitor_pass_status"] = np.random.choice(
    ["Approved", "Pending", "Rejected"],
    size=len(df),
    p=[0.68, 0.27, 0.05]
)

# -----------------------------
# Society Security Level
# -----------------------------
df["society_security_level"] = np.random.choice(
    ["Open", "Moderate", "Strict"],
    size=len(df),
    p=[0.20, 0.45, 0.35]
)

# -----------------------------
# Gate Wait Time (minutes)
# -----------------------------
gate_wait = []

for status in df["visitor_pass_status"]:
    if status == "Approved":
        gate_wait.append(np.random.randint(1, 5))
    elif status == "Pending":
        gate_wait.append(np.random.randint(8, 20))
    else:
        gate_wait.append(np.random.randint(15, 35))

df["gate_wait_time"] = gate_wait

# -----------------------------
# Driver Status
# -----------------------------
df["driver_status"] = np.random.choice(
    ["Available", "Delayed", "Accident", "Medical Emergency"],
    size=len(df),
    p=[0.87, 0.10, 0.02, 0.01]
)

# -----------------------------
# Previous Failed Deliveries
# -----------------------------
df["previous_failed_deliveries"] = np.random.poisson(1, len(df))
df["previous_failed_deliveries"] = df["previous_failed_deliveries"].clip(0, 5)

# -----------------------------
# Address Confidence
# -----------------------------
df["address_confidence"] = np.random.randint(55, 101, len(df))

# -----------------------------
# Preferred Delivery Slot
# -----------------------------
df["preferred_delivery_slot"] = np.random.choice(
    ["Morning", "Afternoon", "Evening"],
    size=len(df),
    p=[0.30, 0.35, 0.35]
)

# -----------------------------
# Estimated Arrival Delay
# -----------------------------
delay = []

for traffic in df["Traffic"]:
    if traffic == "Jam":
        delay.append(np.random.randint(20, 60))
    elif traffic == "High":
        delay.append(np.random.randint(10, 30))
    else:
        delay.append(np.random.randint(0, 15))

df["estimated_arrival_delay"] = delay

# -----------------------------
# Driver Experience
# -----------------------------
df["driver_experience"] = np.random.randint(1, 16, len(df))

# -----------------------------
# Delivery Failure Target
# -----------------------------
failure = []

for _, row in df.iterrows():

    score = 0

    if row["customer_answered_call"] == "No":
        score += 2

    if row["visitor_pass_status"] == "Pending":
        score += 2

    if row["visitor_pass_status"] == "Rejected":
        score += 3

    if row["gate_wait_time"] > 15:
        score += 1

    if row["driver_status"] != "Available":
        score += 2

    if row["previous_failed_deliveries"] >= 2:
        score += 1

    if row["address_confidence"] < 70:
        score += 2

    if row["estimated_arrival_delay"] > 20:
        score += 1

    if row["customer_availability"] == "Travelling":
        score += 2

    failure.append(1 if score >= 5 else 0)

df["delivery_failure"] = failure

# -----------------------------
# Save Dataset
# -----------------------------
df.to_csv("../data/gated_delivery_dataset.csv", index=False)

print("Dataset Updated Successfully!")
print(df.head())
print(df["delivery_failure"].value_counts())
```
