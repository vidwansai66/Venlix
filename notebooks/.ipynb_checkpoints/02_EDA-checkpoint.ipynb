```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

# Display settings
pd.set_option("display.max_columns", None)

# Plot settings
plt.style.use("seaborn-v0_8")
sns.set_palette("Set2")

print("Libraries Imported Successfully ✅")
```

```python
df = pd.read_csv("../data/gated_delivery_dataset.csv")
df.head()
```

```python
print("Dataset Shape:", df.shape)
print("\n")

df.info()
```

```python
df.describe().T
```

```python
missing = pd.DataFrame({
    "Missing Values": df.isnull().sum(),
    "Percentage": (df.isnull().sum()/len(df))*100
})

missing
```

```python
df.dtypes
```

```python
plt.figure(figsize=(6,5))

sns.countplot(x="delivery_failure", data=df)
plt.title("Delivery Failure Distribution")
plt.xlabel("Failed")
plt.ylabel("Count")

plt.show()
```

```python
plt.figure(figsize=(6,4))

sns.countplot(
    x="delivery_failure",
    data=df
)

plt.title("Delivery Failure Distribution")
plt.show()
```

```python
numerical = [
    "Agent_Age",
    "Agent_Rating",
    "Delivery_Time",
    "customer_response_time",
    "gate_wait_time",
    "previous_failed_deliveries",
    "address_confidence",
    "estimated_arrival_delay",
    "driver_experience"
]

df[numerical].hist(
    figsize=(15,10),
    bins=20,
    edgecolor="black"
)

plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(2,3, figsize=(16,8))

cols = [
    "Agent_Age",
    "Agent_Rating",
    "Delivery_Time",
    "customer_response_time",
    "gate_wait_time",
    "previous_failed_deliveries",
    "address_confidence",
    "estimated_arrival_delay",
    "driver_experience"
]

for ax, col in zip(axes.flatten(), cols):

    sns.boxplot(x=df[col], ax=ax)

    ax.set_title(col)

plt.tight_layout()

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.countplot(
    x="Weather",
    hue="delivery_failure",
    data=df
)

plt.xticks(rotation=45)
plt.title("Weather vs Delivery Failure")
plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.countplot(
    x="Traffic",
    hue="delivery_failure",
    data=df
)

plt.title("Traffic vs Delivery Failure")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.countplot(
    x="Vehicle",
    hue="delivery_failure",
    data=df
)

plt.title("Vehicle vs Delivery Failure")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.countplot(
    x="Area",
    hue="delivery_failure",
    data=df
)

plt.title("Area vs Delivery Failure")

plt.show()
```

```python
plt.figure(figsize=(7,5))

sns.boxplot(
    x="delivery_failure",
    y="Agent_Rating",
    data=df
)

plt.title("Driver Rating vs Delivery Failure")

plt.show()
```

```python
plt.figure(figsize=(7,5))

sns.boxplot(
    x="delivery_failure",
    y="Delivery_Time",
    data=df
)

plt.title("Delivery Time vs Delivery Failure")

plt.show()
```

```python
plt.figure(figsize=(7,5))

sns.boxplot(
    x="delivery_failure",
    y="gate_wait_time",
    data=df
)

plt.title("Gate Wait Time vs Delivery Failure")

plt.show()
```

```python
numeric_df = df.select_dtypes(include=np.number)

plt.figure(figsize=(15,10))

sns.heatmap(
    numeric_df.corr(),
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap")

plt.show()
```

```python
sample_df = df.sample(min(3000, len(df)), random_state=42)

pairplot_columns = [
    "Agent_Rating",
    "Delivery_Time",
    "gate_wait_time",
    "address_confidence",
    "estimated_arrival_delay",
    "delivery_failure"
]

sns.pairplot(
    sample_df[pairplot_columns],
    hue="delivery_failure"
)

plt.show()
```

```python
plt.figure(figsize=(7,5))

sns.histplot(
    data=df,
    x="risk_score",
    kde=True,
    bins=20
)

plt.title("Risk Score Distribution")

plt.show()
```

```python
plt.figure(figsize=(7,5))

sns.histplot(
    data=df,
    x="order_value",
    kde=True,
    bins=30
)

plt.title("Order Value Distribution")

plt.show()
```

```python
corr = numeric_df.corr()["label_failed"].sort_values(ascending=False)

print(corr)
```

## EDA Summary

- Dataset contains 43,739 rows and 29 columns.
- No missing values.
- Delivery failures account for approximately 7.4% of the data.
- Higher delivery times are associated with more failures.
- Higher risk scores correlate positively with failed deliveries.
- Weather and traffic conditions influence delivery outcomes.
- Distance shows some outliers that should be reviewed before modeling.
