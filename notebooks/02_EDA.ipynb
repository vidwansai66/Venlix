```python
plt.figure(figsize=(6,5))

sns.countplot(x="label_failed", data=df)

plt.title("Delivery Failure Distribution")
plt.xlabel("Failed")
plt.ylabel("Count")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.countplot(
    y="label_failure_type",
    data=df,
    order=df["label_failure_type"].value_counts().index
)

plt.title("Failure Type Distribution")

plt.show()
```

```python
numerical = [
    "Agent_Age",
    "Agent_Rating",
    "Delivery_Time",
    "distance_km",
    "risk_score",
    "order_value"
]

df[numerical].hist(
    figsize=(15,10),
    bins=20,
    edgecolor="black"
)

plt.tight_layout()

plt.show()
```

```python
fig, axes = plt.subplots(2,3, figsize=(16,8))

cols = [
    "Agent_Age",
    "Agent_Rating",
    "Delivery_Time",
    "distance_km",
    "risk_score",
    "order_value"
]

for ax, col in zip(axes.flatten(), cols):

    sns.boxplot(x=df[col], ax=ax)

    ax.set_title(col)

plt.tight_layout()

plt.show()
```
